```python
## Array - python's array module provides an array that stores elements
#  of the same data type.
''' 
'i'	Signed integer
'f'	Float
'd'	Double/float
A Python list can contain elements of different types, while array.array 
stores elements of a single specified type and is generally more 
memory-efficient for homogeneous numeric data.
'''
import array
arr=array.array('i',[1,2,3,4])  ## 'i' is the type code specifying that the array contains integers.
print(arr)
```

```python
''' Some important math functions: 
math.ceil(4.2)     # 5
math.floor(4.8)    # 4
math.factorial(5)  # 120
math.pow(2, 3)     # 8.0
math.gcd(12, 18)   # 6
math.log(10)       # Natural logarithm

math.pow() returns a float, while ** is Python's exponentiation 
operator and preserves an appropriate numeric type.
'''

import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random
'''
randint → random integer inclusive of both ends
choice → random element from a sequence.

random.random()       # float between 0.0 and 1.0
random.randrange(1, 10) # random integer from 1 to 9
random.shuffle(lst)   # randomly rearranges a list
'''
import random
print(random.randint(1,10))  ## gives a random integer  and LB and UB both included
print(random.choice(['apple','banana','cherry']))  ## gives a random choice
```

```python
## High level operations on files and collection of files
'''
shutil = shell utilities. It provides high-level operations for 
copying, moving, and deleting files/directories.

shutil.copyfile(src, dst)       # Copy file contents
shutil.copy(src, dst)           # Copy file + permissions
shutil.copy2(src, dst)          # Copy file + metadata
shutil.move(src, dst)           # Move/rename
shutil.rmtree("folder")         # Delete directory recursively

os → interact with the operating system
shutil → high-level file/directory operations
'''
import shutil
shutil.copyfile('source.txt','destination.txt')
```

```python
## Data Serialization
import json
data={'name':'Krish','age':25}

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## CSV

import csv
with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['krish',32])
```

```python
with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```
