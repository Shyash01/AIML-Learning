Importing Modules in Python: Modules And Packages   

```python
# math is a function -> specifically mathematica operations
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
from math import *  ## Import all the modules inside maths
print(sqrt(16))
print(pi)
```

```python
# Custom packages
## 1. Created a folder named package for making custom packages
## 2. Create a file __init__.py inside packages
## 3. Create the package file
from package.maths import addition

addition(2,3)
```

```python
from package import maths
maths.addition(2,3)

maths.substraction(3,5)
```
