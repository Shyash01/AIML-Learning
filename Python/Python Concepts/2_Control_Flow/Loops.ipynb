```python
import math

num = int(input("Enter number: "))
print(f"Prime number upto {num} are: ")
if num < 1:
    print("invalid input")
elif num == 1:
    print("No Prime Exists")
elif num == 2:
    print(f"{num} is a prime number")
else:
    print("2 is a prime number")
    # Use num + 1 if you want to include the entered number in the check
    for i in range(3, num + 1): 
        
        if i % 2 == 0:
            continue
        elif i > 1:
            isPrime = True # 1. RESET FLAG HERE for every new number
            
            limit = int(math.isqrt(i)) # 2. CHANGE TO i instead of num
            
            # limit + 1 ensures the square root value itself is checked (like 3 for 9)
            for j in range(3, limit + 1, 2): 
                if i % j == 0:
                    isPrime = False
                    break
            if isPrime:
                print(f"{i} is a prime number")
```
