
### try	
Wraps the risky code that might throw an error -> Always runs first.
### except	
Catches and handles specific exceptions. -> Runs only if an error occurs inside try.
### else	
Separates the normal flow from error handling. -> Runs only if no exceptions occur in try.
### finally	
Used for cleanup (e.g., closing databases or files). -> 
Always runs, regardless of whether an error occurred.





```python
try:
    # Code that might raise an exception
    file = open("data.txt", "r")
    content = file.read()
    value = 10 / int(content.strip())
except FileNotFoundError:
    print("Error: The requested file does not exist.")
except ZeroDivisionError:
    print("Error: The file contains a zero; cannot divide by zero.")
except ValueError as e:
    print(f"Error: Invalid data format. Details: {e}")
else:
    print(f"Operation successful! Result is: {value}")
finally:
    # This block executes no matter what happens above
    print("Cleaning up resources...")
    try:
        file.close()
    except NameError:
        pass # File was never opened
```
