Lambda Function -> anaonymous function -> lambda keyword
- can have any number of arguments but only one expression
- commonly used for short operations -> higher order functions

```python
# Syntax

lambda arguments: expression
```

```python
def addition(a,b):
    return a+b

addition(3,4)
```

```python
addition2 = lambda a,b:a+b
print(type(addition))
print(addition2(5,6))
```

```python
def checkEven(num):
    if num%2==0:
        return True


checkEven(24)
```

```python
checkEven2 = lambda num :  num%2==0

checkEven2(24)
```

```python
def addition(x,y,z):
    return x+y+z

addition(2,3,4)
```

```python
addition3= lambda a,b,c:a+b+c
addition3(4,3,2)
```

```python
addition4 = lambda a,b:a/b + a%b
addition4(9,2)
```

```python
## Map() => applies a function to all items in a list/collection

numbers = [1,2,3,4,5,6]
def square(number):
    return number**2

square(2)
```

```python
## Map (function-expression,iterable)
list(map(lambda x:x**2,numbers))
```
