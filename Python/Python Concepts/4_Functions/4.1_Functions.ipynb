```python
### Variable Length Arguments
## Positional And Keyword Arguments

## We can give any number of parameters 
# *args -> positional argument -> we can keep *krish, *yash -> any name 
def print_numbers(*Yash):
    for number in Yash:
        print(number)
    
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Krish")  ## 1,2,3,... All are positional parameters
# *Yash will be refering these parameters.
```

```python
### Keywords Arguments
## **kwards -> args as positional argument forming -> keyword argument
# In keywords Arguments -> all the parameters in form key value pairs
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")


    
```

```python
print_details(name="Krish",age="32",country="India")
```

```python
print_details(12,22,Yash) 

# Positional arguments take single value
# Keyword arguments take only key value pairs
```

```python
def print_details2(*args,**kwargs):

    for val in args:
        print(f"{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details2(1,2,3,name="Yash")   # If making mix of positional and keyword argument 
# Then keep in order as defined in above function 
#Positional argument cannot appear after keyword arguments
```

```python
def print_details3(**kwargs,*args):  ##Parameter cannot follow ** 

    for val in args:
        print(f"{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```
