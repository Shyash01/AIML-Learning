### MAP FUNCTION

map() -> applies given function to all items -> input list
return a map object(an iterator)
useful for transforming data in a list comprehensively

```python
def square(x):
    return x*x

square(10)
```

```python
numbers=[1,2,3,4,5,6,7,8]
list(map(square,numbers))  ## two input parameters -> function and *iterable(iterable is positional argument)
```

```python
## Lambda function with map
numbers=[1,2,3,4,5,6,7,8]
list(map(lambda x:x*x,numbers))
```

```python
## Map Multiple iterables
numbers1=[1,2,3]
numbers2=[4,5,6]

added_List=list(map(lambda a,b:a+b,numbers1,numbers2))
print(added_List)
```

```python
## Map to convert list of strings to integers
# Use map to convert strings to integers
str_numbers = ['1','2','3','4','5']
int_numbers = list(map(int,str_numbers))
print(int_numbers)
```

```python
words=['apple','banana','orange']
upper_word=list(map(str.upper,words))
print(upper_word)
```

```python
def get_name(person):
    return person['name']

people=[
    {'name':'Krish','age':32},
    {'name':'Jack','age':33}
]

list(map(get_name,people))
```
