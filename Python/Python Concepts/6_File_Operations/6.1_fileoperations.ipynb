```python
# File operations -> r/w files
```

```python
## Read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
# Read a file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())  ## ssstrip() removes the newline character
```

```python
## Writing a file(Overwriting)

with open('example.txt','w') as file:
    file.write("Helllooo..People\n")
    file.write("This is the new line.\n")
    ## Removes existing content and then updates with the new content
 
```

```python
## Write a file(without overwriting)

with open('example.txt','a') as file:
    file.write('This is the new line to be appended\n')
```

```python
## Writing a list of lines to a file
lst = ["This is first line\n",
       "This is second line\n",
       "This is third line"
       ]

with open('example.txt','a') as file:
    file.writelines(lst)

```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) # strip() -> removes newline character
```

```python
### Binar Files

# writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
# Reading a binary file
with open('example.bin','rb') as file:
    content=file.read()
    print(content)

```

```python
## Read the content from a source text file and 
## write to a destionation text file

# Copying a text file
with open('source_file.txt','r') as source_file:
    content=source_file.read()

with open('destination_file.txt','w') as destination_file:
    destination_file.write(content)

with open('destination_file.txt','r') as destination_file:
    content=destination_file.read()
    print(content)
```

```python
## w+ mode -> open file in both read and write mode
# if file not exsits -> created 
# if exists -> contents gets truncated i.e. Overwritten

with open('example.txt','w+') as file:
    file.write('Checking w+ mode\n')
    file.write('Second line for checking\n')

    ## Move the file cursor to beginning
    file.seek(0)

    ## Read the content of the file
    content=file.read()
    print(content)
```
