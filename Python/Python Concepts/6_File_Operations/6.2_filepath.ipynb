```python
## Using Os module
import os

cwd = os.getcwd()
print(f"The current working directory is: '{cwd}'")
```

```python
## Create a new directory
import os

new_directory1 = "package2"
os.mkdir(new_directory1)

```

```python
## Listing all files and directories
lst=os.listdir('.')
print(lst)
```

```python
## Joining paths
file_name="example.txt"
dir_name="package"
full_path=os.path.join(dir_name,file_name)
print(full_path)
```

```python
## Joining path with cwd

dir_name="folder"
file_name="file.txt"
full_path=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
# Checking if a Path exists
import os

path='example1.txt'

if os.path.exists(path):
    print(f"Given path {path} exists")
else:
    print(f"Path {path} doesn't exists")
```

```python
# Checking if a path is a file or directory

path = 'example2.txt'

if os.path.isfile(path):
    print(f"Given path {path} is a file")
elif os.path.isdir(path):
    print(f"Given path {path} is a directory")
else:
    print(f"Given path is neither file nor directory")
```

```python
## Getting the absolute path
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```
