**Question 5:** Write a Python program to calculate the factorial of a number.

```python
# Your code here
n = int(input("Enter number: "))

fact = 1

if n==0 or n==1:
    print(1)
else:
    for i in range(1,n+1):
        fact *=i

print(fact)
```

**Question 8:** Write a Python program to convert Celsius to Fahrenheit.

```python
# Your code here
def tempConvert(temp,unitToConvert):
    
    if unitToConvert.lower()=='c':
        return temp*(5/9) - 32
    elif unitToConvert.lower()=='f':
        return temp*(9/5) + 32
    else:
        return "check credentials"

temp=int(input("Enter temperature: "))
unitToConvert=input("Enter unit in which conversion is required(c,f): ")

print(tempConvert(temp,unitToConvert))
    
```
