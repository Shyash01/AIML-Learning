Dictionaries -> unordered collection of items -> store data in k,v pairs -> keys must be unique and immutable -> values can be any and of any type

-> Key can be either string number or tuple

```python
# Creating Dictionaries
empty_Dict={}
print(empty_Dict)
print(type(empty_Dict))
```

```python
emp_Dict=dict()   ## Using dict() class to create dictionary
emp_Dict
```

```python
student={"name":"Yash","age":22,"grade":9.27,"isActive":True}
print(student)
```

```python
# Key should be unique -> if duplicate -> last instance of value is picked as final
# for that very specificy key

student={"name":"Yash","age":22,"grade":9.27,"isActive":True,"name":"Raunak","age":21,"grade":8.27,"isActive":False,"isPlace":True}
print(student)
```

```python
## Accessing dictionary elements
print(student['name'])
print(student['isActive'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('isMarried'))
print(student.get('isPlace'))
print(student.get('isMarried',"Not Married"))
print(student.get('isActive',"Bhaag gaya saala"))
```

```python
## Modifying dictionary elements
## Dictionaries are mutable -> you can add, update or delete elements
## But only keys are mutable

print(student)
```

```python
student["age"] = 24
print(student)

student["address"] = "Gurugram"  ## Updates a value of key if exists otherwise create new key value pair
print(student)
```

```python
del student['grade']

print(student)
```

```python
## Dictionary Methods

keys=student.keys()  ## Get all Keys
print(keys)

values = student.values()
print(values)

items=student.items()
print(items)
```

```python
## Shallow copy
student_copy = student  ## Normal Copy
print(student)
print(student_copy)
```

```python
student["name"] = "krish2"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy()
print(student_copy1)
print(student)
```

```python
student["name"]="Krish3"
print(student_copy1)
print(student)
```

```python
## Iterating over Dictionaries
## You can use loops to iterate over dictionaries, keys, values or items

## Iterating over keys: 

for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for value in student.values():
    print(value)
```

```python
## Iterate over key value pairs

for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Dictionaries
students={
    "student1":{"name":"Krish","age":32},
    "student2":{"name":"Peter","age":35}
}

print(students)
```

```python
## Access Nested Dictionaries Elements

print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comphrehension
squares = {x:x**2 for x in range(0,5)}
print(squares)

print("\n")

dictionary={}
for key in range(0,5):
    dictionary[key]=key*key

print(dictionary)
```

```python
## Condition dictionary comprehension
evenSquare={x:x**2 for x in range(10) }
print(evenSquare)
```

```python
## Practical Examples

## Use a dictionary to count the frequency of elements in list

numbers=[1,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge 2 dictionaries into one

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}

merged_dict={**dict1,**dict2}
print(merged_dict)
```
