```python
# Q1
s = set()
s1 = set(range(1,11))
print(s1)
```

```python
s1.add(11)
print(s1)
print(s1.pop())  # pop doesnt take argument
print(s1)
```

```python
s2=set(range(1,6))
s3=set(x for x in range(1,11) if x%2==0)
print(s2)
print(s3)

su=s2.union(s3)
print(su)

si = s2.intersection(s3)
print(si)

sd = s2.difference(s3)
print(sd)

ssd = s2.symmetric_difference(s3)
print(ssd)
```

```python
print(s2 | s3)
print(s2 & s3)
print(s2-s3)
print(s2 ^ s3)
```
