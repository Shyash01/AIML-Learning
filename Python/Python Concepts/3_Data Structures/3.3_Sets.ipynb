     -> Unordered

```python
## Create a set
mySet = {1,2,3,4,5}
print(mySet)
print(type(mySet))
```

```python
## Using set() class to initialise an empty set

empty_Set = set()
print(type(empty_Set))
```

```python
set1=set([1,2,3,5,4,6,5,6,])  ## Using List to initialise a set
print(set1)     ## Only unique elements are printed with sorted order
```

```python
set2={1,2,3,'a','A','$','3'}  ## Hybrid Set
print(set2)
```

```python
# Basic sets operation
# Adding and removing elements
print(set1)
```

```python
set1.add(6)
print(set1)
set1.add(6)
print(set1)
```

```python
# Remove the elements from a set
set1.remove(5)   ## Rermove : If try to remove non existing value -> throws an error
print(set1)
```

```python
set1.remove(10)
```

```python
set1.add(11)
print(set1)
```

```python
set1.discard(11)  ## discard : doesn't throw error if try to remove non exisiting value
print(set1)  ## remove the element if its a member, does not raise an exception -> when element missing
```

```python
# pop method
print(set1)
```

```python
removed_ele = set1.pop()  ## Pops the first element from the set 
print(removed_ele)
print(set1)
```

```python
## Clear all the elements
set1.clear()
print(set1)
```

```python
# Set Membership test
set2 = {1,2,3,4,5}
print(3 in set2)
print(6 in set2)
```

```python
# Mathematical Operation
set1 = {1,3,4,5,6,7,9}
set2 = {2,3,4,6,7,8,10}

## Union
union_set = set1.union(set2)
print(union_set)

## Intersection
inter_set = set1.intersection(set2)
print(inter_set)

## intersection update
set1.intersection_update(set2)  ## Updates the set1 with intersection result
```

```python

print(set1)  ## set1 result after intersection_update
```

```python
# Mathematical Operation
set1 = {1,3,4,5,6,7,9}
set2 = {2,3,4,6,7,8,10}

print(set1.difference(set2))
```

```python
## Set Methods
set1 = {1,2,3,4,5}
set2 = {3,4,5}

# is subset
print(set1.issubset(set2))
print(set2.issubset(set1))
print(set1.issuperset(set2))
```

```python
# Counting unique words in text
text = "This is the tutorial about sets, is this the tutorial"
words = text.split()   ## Splits each word in text by space
print(words)

## Converting list of words to set to get unique words
uniqueWords = set(words)
print(uniqueWords)
print(len(uniqueWords))
```
