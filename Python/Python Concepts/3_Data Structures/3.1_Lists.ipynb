```python
# 2. Modifying fruits2 Changed fruits
"""
The line fruits2 = fruits does not create a copy of your list. 
Instead, it creates a new reference pointing to the exact same list
in memory. Because they share the same identity, modifying fruits2 
automatically updates fruits. That is why printing fruits in the second 
cell shows the modified character list.
"""
fruits2=fruits
print(fruits2)
```

## Enumerate
The enumerate() function -> built-in tool that allows you to loop through an iterable (like a list, tuple, or string) while automatically keeping track of the index (counter) and the element value at the same time.
This eliminates the need to manually create and increment a counter variable outside your loop.

### Basic Syntax
enumerate(iterable, start=0)

```python
## Iterating with index -> Using enumerate
for index,i in enumerate(numbers):
    print(index,i)
```

```python
## Iterating with index -> Using enumerate
for index,i in enumerate(numbers,5):
    print(index,i)
```

```python
## List Comprehension
lst=[]
for x in range(10):
    lst.append(x**2)
print(lst)
```

```python
[x**2 for x in range(12)]
```

```python
## List Comprehension with condition
lst=[]
for i in range(10):
    if i%2==0:
        lst.append(i)
print(lst)
```

```python
even_Lst = [num for num in range(10) if num%2==0]
print(even_Lst)
```

```python
even_lst=[num for num in range(10) if num%2!=0]
print(even_lst)
```

```python
[x**2 for x in range(10) if x%2==0]
```

```python
## Nested List Comprehension
lst1 = [1,2,3,4]
lst2 = ['a','b','c','d']

pair=[[i,j] for i in lst1 for j in lst2]
print(pair)
```

```python
## List Comprehension with function calls
words=["hello","world","python","list","comprehension"]
lengths=[len(word) for word in words]
print(lengths)
```
