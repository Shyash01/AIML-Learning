Real World Examples Using Lists in Python

```python
##Example 1 - Manage A to do list

# Create a To Do List To Keep Track Of Tasks

to_do_list=["Buy Groceries","Clean the house","Pay Bills"]

## Adding a task in to do list:
to_do_list.append("Schedule meeting")
to_do_list.append("Wash Clothes")

## Removing a cmpleted task
to_do_list.remove("Clean the house")

## Checking if a task is in the list

if "Wash Clothes" in to_do_list:
    print("Dont forget to wash clothes")

print("To do list remaining: ")
for task in to_do_list:
    print(f"-{task}")
```

```python
## Create a to do list: Using Dictionary

to_do_list2={}
taskNo = 0
val = input("Press y to open and n to close to do list")
print("Enter Options: \n1 - Add a task\n2 - Remove a task\n3 - Print Current Task")
print(to_do_list2)
while(val == 'y'):

    
    choice=int(input("Enter choice: "))

    if choice==1:
        task=input("Enter Task (50 words maximum): ")
        taskNo=taskNo+1
        to_do_list2[taskNo] = task
        
    elif choice==2:
        comp = int(input("Enter completed task number: "))
        del to_do_list2[comp]

    elif choice==3:
        for taskNo,task in to_do_list2.items():
            print(f"{taskNo}:{task}")

    else: 
        print("Enter Options: \n1 - Add a task\n2 - Remove a task\n3 - Print Current Task")

    val = input("Press y to open and n to close to do list")
        

```

```python
## Example 2: Organising student Grades

grades = [34,36,89,92,93]

grades.append(94)

average_grade = sum(grades)/len(grades)
print(f"Average Grade: {average_grade:.2f}")

highest_grade = max(grades)
lowest_grade = min(grades)

print(f"Highest Grade: {highest_grade}")
print(f"Lowest Grade: {lowest_grade}")
```
