```python
## Packing and Unpacking tuple
## Packing
packed_tuple=1,"hello",3.14  ## If elements given without braces () or [] and comma seprated
                             ## the elements are packed as tuple
print(packed_tuple)
```

```python
## Unpacking a tuple 
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking a tuple 
a,b,c,d=packed_tuple
print(a)
print(b)
print(c)
```

```python
## Nested Tuples

tpl = ((1,2,3),('a','b','c'),("Yash",2,2.12,True))
print(tpl)
```

```python
## Nested Tuples

tpl = ((1,2,3),('a','b','c'),("Yash",2,2.12,True))
tpl[2][1:]
```

```python
## Iterating over nested tuples
for i in tpl:
    for j in tpl:
        print(i,j)
```

```python
## Iterating over nested tuples
for i in tpl:
    for j in i:
        print(j)
```

```python
## Iterating over nested tuples
nested_tuple = ((1, 2, 3), ("a", "b", "c"), (True, False))

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
