```python
#Q15

tp8=(1,2,5,7,9,2,0)
def min_in_tuple(tp8):
    return min(tp8)

def max_in_tuple(tp8):
    return max(tp8)

def sum_in_tuple(tp8):
    return sum(tp8)

print(min_in_tuple(tp8))
print(max_in_tuple(tp8))
print(sum_in_tuple(tp8))
```
