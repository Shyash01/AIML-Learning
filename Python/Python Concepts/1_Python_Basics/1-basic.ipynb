Syntax and Semantics in Python

```python
## Basic Syntax Rules
## Case sensitivity 
name = "Krish"
Name = "Naik"

print(name)
print(Name)
```

```python
## Indentation

age = 30
if age>30:
    print(age)
print(age)
```

```python
## Line Continuation
## Use (\) -> continue to next line

total = 1+2+3+4+\
4+5+6

print(total)
```

```python
## Multiple statements on a single line
x =5;y=10;z=x+y
print(z)
```

```python
## Understand Semantics in Python
# Variable Assignment
age = 32 ## Age is an integer
name = "Krish" ## name is a string
```

```python
## type inference -> enforce type at runtime, allow dynamic typing
variable = 10
print(type(variable))
variable = "Krish"
print(type(variable))  ## During runtime , type can be changed.
```
